```python
#import libraries
import pandas as pd
import numpy as np
import scipy.signal as sci
import matplotlib.pyplot as plot
```

```python
# TXT-Datei einlesen
df = pd.read_csv('ACC_Data_Aufgabe1.4.txt', sep='\t', header=None,
                 names=['X', 'Y', 'Z', 'Time'])

# Zeit relativ zum ersten Messpunkt in Sekunden umrechnen
df['Time'] = (df['Time'] - df['Time'].iloc[0]) / 1000
```

```python


# Alle drei ACC-Achsen über die ersten vier Sekunden plotten
df_4s = df.loc[df['Time'] < 4]
df_4s.plot(
    x='Time',
    y=['X', 'Y', 'Z'],
    ylabel='Beschleunigung (Messwerte)',
    xlabel='Zeit (s)',
    figsize=(10, 5),
    title='ACC-Daten – erste 4 Sekunden'
)

plot.xlim(0, 4)
plot.grid(True)
plot.tight_layout()
plot.show()
```
